##### **7. Detección de Valores Atípicos (Outliers)**

✔ Identificar valores que se alejan significativamente del resto.

✔ Aplicar métodos estadísticos para encontrar valores extremos.

✔ Evaluar si el outlier es un error o un dato válido.


### **Métodos recomendados**

#### **a) Método del Rango Intercuartílico (IQR)**

Detecta valores por debajo de **Q1 – 1.5*IQR** y por encima de **Q3 + 1.5*IQR**.

**Código base:**

```python
Q1 = df['columna'].quantile(0.25)
Q3 = df['columna'].quantile(0.75)
IQR = Q3 - Q1

outliers = df[(df['columna'] < Q1 - 1.5*IQR) | (df['columna'] > Q3 + 1.5*IQR)]
```

#### **b) Método de la Desviación Estándar**

Valores mayores a **media ± 3*desviación estándar**.

**Código base:**

```python
media = df['columna'].mean()
std = df['columna'].std()

outliers = df[(df['columna'] < media - 3*std) | (df['columna'] > media + 3*std)]
```

* `df.boxplot(column='col')`
* `df['col'].plot(kind='box')`

# **8. Tratamiento de Outliers**

✔ Analizar si el outlier debe **eliminarse**, **corregirse** o **reemplazarse**.

✔ Evaluar si el outlier es un error humano, un error del sistema o un valor real extremo.

### Opciones de tratamiento:

#### **a) Eliminar outliers**

```python
df = df[(df['col'] >= limite_inferior) & (df['col'] <= limite_superior)]
```

#### **b) Reemplazar por valores estadísticos**

* Media
* Mediana
* Percentiles
* Valor máximo permitido
* Winsorización (capar valores extremos)

```python
df['col'] = np.where(df['col'] > limite_superior, limite_superior, df['col'])
```

#### **c) Corrección manual**

Si se detecta un error claro (ej: edad = 450), reemplazar con criterio.

# **9. Funciones de Imputación (Pandas)**

✔ Rellenar valores faltantes con métodos estadísticos o basados en el dataset.

### **Funciones útiles en pandas:**

#### **a) Imputar con media**

```python
df['col'] = df['col'].fillna(df['col'].mean())
```

#### **b) Imputar con mediana**

```python
df['col'] = df['col'].fillna(df['col'].median())
```

#### **c) Imputar con moda**

```python
df['col'] = df['col'].fillna(df['col'].mode()[0])
```

#### **d) Imputación hacia adelante o hacia atrás**

```python
df['col'] = df['col'].fillna(method='ffill')
df['col'] = df['col'].fillna(method='bfill')
```

#### **e) Imputación condicional**

Completar usando información de otra columna:

```python
df.loc[df['ciudad'] == "Arroyito", "edad"] = df['edad'].median()
```

---

# **10. Criterios para seleccionar el método adecuado de imputación**

### **1. Imputar por *media* → cuando:**

✔ Distribución normal (simétrica)

✔ Datos numéricos continuos

✔ No hay muchos valores extremos


Ejemplo: **edad** (si no tiene outliers extremos)

---

### **2. Imputar por *mediana* → cuando:**

✔ Hay outliers presentes

✔ Distribución sesgada

✔ Variable numérica como ingresos, gastos, montos


Ejemplo: **ingresos**, **importe_final_orden**

---

### **3. Imputar por *moda* → cuando:**

✔ La variable es **categórica**

Ejemplos:

* sexo
* ciudad
* nivel_estudio
* situacion_laboral

---

### **4. Imputación usando otros registros del dataset → cuando:**

✔ La columna está relacionada con otra

✔ El valor faltante se puede deducir por patrones internos

✔ Se puede completar usando un subconjunto de datos


Ejemplos:

* Si **ciudad** está vacía → usar **filas con el mismo id_provincia**
* Si **nivel_estudio** falta → usar registros con **edad similar**
* Si falta **situacion_laboral** → mirar su **trayectoria o estudios**

**Ejemplo de imputación condicional:**

```python
df.loc[df['edad'] > 30, 'nivel_estudio'] = df['nivel_estudio'].mode()[0]
```

````python
import pandas as pd

# Recreamos el DataFrame df_ventas para la demostración
ventas_dirty_data = {
    'ID_Transaccion': [1, 2, 2, 4, 5, 6, 7, 8],
    'Fecha': ['2024-01-10', '11/01/2024', '11/01/2024', '2024-01-13', '2024.01.14', '15-Jan-2024', None, '2024-01-17'],
    'Cliente': [' Juan Perez ', 'MARIA GOMEZ', 'MARIA GOMEZ', 'Pedro perez', 'Ana Lopez', 'Carlos Ruiz', 'Laura Fernandez', '  juan perez'],
    'Monto_USD': ['150.50', '200.00', '200.00', '350.00', '0.00', '180.00', '500.50', '150.50'],
    'Estado': ['Completado', 'completado', 'completado', 'P_Pendiente', 'Completado', 'CANCELADO', 'Completado', 'Completado']
}
df_original = pd.DataFrame(ventas_dirty_data)

print("DataFrame Original:")
print(df_original)
print("\n" + "-"*30 + "\n")

# Caso 1: SIN usar .copy()
df_sin_copy = df_original.drop_duplicates()
print("DataFrame después de drop_duplicates() SIN .copy():")
print(df_sin_copy)

# Intentamos modificar df_sin_copy (esto podría generar SettingWithCopyWarning)
# Para forzar un cambio visible, vamos a modificar un valor.
# Nota: El warning no siempre aparece, pero el problema subyacente de vista vs copia persiste.
print("\nIntentando modificar df_sin_copy (fila 0, columna 'Estado'):")
df_sin_copy.loc[0, 'Estado'] = 'MODIFICADO_SIN_COPY'

print("\ndf_sin_copy después de la modificación:")
print(df_sin_copy)
print("\n¿Cómo afectó esto al df_original?")
print(df_original) # A veces se modifica, a veces no, dependiendo de la optimización interna de Pandas
print("\n" + "-"*30 + "\n")

# Reiniciamos df_original para el siguiente caso
df_original = pd.DataFrame(ventas_dirty_data)

# Caso 2: USANDO .copy()
df_con_copy = df_original.drop_duplicates().copy()
print("DataFrame después de drop_duplicates() CON .copy():")
print(df_con_copy)

# Intentamos modificar df_con_copy
print("\nIntentando modificar df_con_copy (fila 0, columna 'Estado'):")
df_con_copy.loc[0, 'Estado'] = 'MODIFICADO_CON_COPY'

print("\ndf_con_copy después de la modificación:")
print(df_con_copy)
print("\n¿Cómo afectó esto al df_original?")
print(df_original) # df_original permanece sin cambios

print("\nConclusión: Como puedes ver en el 'Caso 2', usar `.copy()` asegura que cualquier modificación en `df_con_copy` no afecte al `df_original`.")
print("En el 'Caso 1', la modificación de `df_sin_copy` *podría* haber modificado también `df_original`, lo cual es un comportamiento no deseado y difícil de rastrear. Por eso es una buena práctica usar `.copy()` cuando se espera una copia independiente.")
````

##### Diferencias entre un dataset de la librería y uno de un repositorio

| Aspecto | `sns.load_dataset()` | `pd.read_csv(url)` |
|---|---|---|
| **Origen** | Repositorio oficial `seaborn-data`, curado por los mantenedores de la librería | Cualquier repositorio público, tal como lo subió su autor |
| **Estado del dato** | **Preprocesado**: nombres en minúscula, columnas derivadas agregadas, tipos `category` ya asignados | **Crudo**, exactamente como se publicó |
| **Columnas (Titanic)** | 15 columnas: agrega `class`, `who`, `adult_male`, `deck`, `embark_town`, `alive`, `alone`; y **elimina** `PassengerId`, `Name`, `Ticket`, `Cabin` | Las 12 columnas originales |
| **Parseo** | Resuelto por la librería | A veces hay que dárselo uno: `iris.data` no tiene encabezado y exige `names=` |
| **Etiquetas (Iris)** | `setosa`, `versicolor`, `virginica` | `Iris-setosa`, `Iris-versicolor`, `Iris-virginica` |
| **Estabilidad** | Versionada junto con la librería; se cachea localmente tras la primera descarga | La URL puede cambiar, moverse o caerse; el contenido puede editarse sin aviso |
| **Reproducibilidad** | Alta: fijando la versión de seaborn se fija el dato | Menor: hay que versionar el archivo por cuenta propia |
| **Uso típico** | Docencia, ejemplos, práctica de visualización | Trabajo real, fuentes originales, práctica de limpieza |

### Conclusión

Ambas opciones requieren conexión a internet (seaborn descarga desde GitHub la
primera vez y después usa caché). La diferencia de fondo es de **propósito**:

- `seaborn` entrega un dataset **ya masticado**, pensado para que el ejemplo de
  gráfico funcione en una línea. Es cómodo, pero al venir limpio **oculta
  justamente el trabajo que esta etapa busca enseñar**.
- El repositorio entrega el **dato original con toda su suciedad**: faltantes,
  formatos inconsistentes, columnas sin nombre.

Por eso, para un ejercicio de exploración y limpieza, la opción correcta es la 2.
Un detalle no menor: `deck` en la versión de seaborn es la misma información que
`Cabin` en la versión cruda, pero ya reducida a la letra de la cubierta. Si uno
trabaja solo con seaborn, nunca ve el problema de datos que había detrás.

| Transaction ID | int64 | Identificador | Debe ser único |
| Date | object | Fecha | Llega como texto, no como `datetime` |
| Customer ID | object | Identificador | Formato "CUSTxxx" |
| Gender | object | Categórica nominal | Male / Female |
| Age | int64 | Numérica discreta | Edad del cliente |
| Product Category | object | Categórica nominal | Beauty / Clothing / Electronics |
| Quantity | int64 | Numérica discreta (conteo) | Unidades vendidas |
| Price per Unit | int64 | Numérica continua | Precio unitario |
| Total Amount | int64 | Numérica continua | Variable derivada: Quantity × Price per Unit |

### 2) ¿Qué impresiones iniciales surgen?

- El dataset **original** está completamente limpio: sin nulos, sin duplicados,
  tipos coherentes y la regla `Total = Quantity × Price` se cumple siempre. Es un
  dataset de ejemplo, no un volcado real de ventas.
- Al generar `df_ventas_duplicadas`, el `.sample(n=num_duplicates, replace=True)`
  toma filas **al azar con reemplazo**, así que puede repetir la misma fila más de
  una vez o duplicar filas distintas entre sí. La cantidad exacta de duplicados
  varía en cada corrida (entre 10 y 20), porque `num_duplicates` es aleatorio.
- El efecto directo es que **`Transaction ID` deja de ser único**: al duplicar una
  fila completa se duplica también su identificador, algo que en un sistema real
  de ventas sería una violación de integridad (dos ventas distintas no pueden
  compartir el mismo ID de transacción).
- Como el duplicado es una copia exacta de una fila válida, **no aparecen nulos
  nuevos ni cambios de tipo**: `isnull().sum()` da igual que en el original. El
  único síntoma es el aumento de filas y de duplicados.

### 3) ¿Qué "cosas raras" aparecen a simple vista?

| Error | Manifestación | Cómo se detecta |
|---|---|---|
| **Filas duplicadas exactas** | Entre 10 y 20 filas repetidas letra por letra | `df.duplicated().sum()` |
| **Identificador repetido** | El mismo `Transaction ID` aparece más de una vez | `df["Transaction ID"].duplicated().sum()` |
| **Alteración del tamaño muestral** | `len(df)` crece sin que haya "ventas nuevas" reales | Comparar `shape` antes/después |
| **Sesgo estadístico potencial** | Si `sample()` repite varias veces la misma fila, esa transacción queda **sobrerrepresentada** en `describe()` y en cualquier promedio o conteo por categoría | Comparar `value_counts()` y `describe()` del original contra el duplicado |

El punto más importante para el informe es este último: un duplicado no es solo
"una fila de más", es un **sesgo silencioso**. Si por azar `sample()` elige varias
veces una venta de `Electronics` de alto monto, el `Total Amount` promedio y la
distribución por categoría se corren sin que haya pasado nada raro a simple vista
en el `head()`.

##### Etapa 3 — Experimentación

Para cada dataset se comprueban las hipótesis de la etapa anterior mediante
código, organizado en las cuatro dimensiones de calidad de datos:

- **Completitud** → `isnull().sum()`
- **Unicidad** → `duplicated()`
- **Validez** → condiciones lógicas sobre rangos o formatos esperados
- **Consistencia** → comparación entre columnas relacionadas

Donde resulta útil, se agrega una **columna auxiliar booleana** que marca fila
por fila si el registro cumple o no la condición evaluada.

````python
# ============================================================
# EXPERIMENTACIÓN — TITANIC
# ============================================================

# ---------- COMPLETITUD ----------
print(">>> COMPLETITUD")
print(df_titanic.isnull().sum())

# Columna auxiliar: ¿tiene cabina registrada? (vinculado a H2 de faltantes por clase)
df_titanic["cabin_registrada"] = df_titanic["Cabin"].notnull()
print("\nCabina registrada por clase (True = tiene dato):")
print(df_titanic.groupby("Pclass")["cabin_registrada"].mean().round(2))
# Confirma H2: la proporción de faltantes en Cabin/Age es mucho mayor en 3ª clase.

# ---------- UNICIDAD ----------
print("\n>>> UNICIDAD")
print("Filas totalmente duplicadas:", df_titanic.duplicated().sum())
print("PassengerId duplicados:", df_titanic["PassengerId"].duplicated().sum())
print("Tickets que se repiten (grupo/familia comparte pasaje):",
      df_titanic["Ticket"].duplicated().sum())

# ---------- VALIDEZ ----------
print("\n>>> VALIDEZ")
df_titanic["edad_valida"] = df_titanic["Age"].between(0, 100) | df_titanic["Age"].isnull()
df_titanic["fare_valido"] = df_titanic["Fare"] >= 0
df_titanic["pclass_valida"] = df_titanic["Pclass"].isin([1, 2, 3])
df_titanic["sexo_valido"] = df_titanic["Sex"].isin(["male", "female"])

print("Edades fuera de 0-100:", (~df_titanic["edad_valida"]).sum())
print("Fare negativo:", (~df_titanic["fare_valido"]).sum())
print("Pclass fuera de {1,2,3}:", (~df_titanic["pclass_valida"]).sum())
print("Sex fuera de {male,female}:", (~df_titanic["sexo_valido"]).sum())

# ---------- CONSISTENCIA ----------
print("\n>>> CONSISTENCIA (H1: Fare por ticket compartido)")
personas_por_ticket = df_titanic.groupby("Ticket")["Ticket"].transform("count")
df_titanic["fare_por_persona"] = df_titanic["Fare"] / personas_por_ticket
print("Fare promedio sin ajustar:", df_titanic["Fare"].mean().round(2))
print("Fare promedio ajustado por persona:", df_titanic["fare_por_persona"].mean().round(2))

# Consistencia adicional: el título del nombre debería coincidir con el sexo
df_titanic["titulo"] = df_titanic["Name"].str.extract(r",\s*([^\.]+)\.")
titulos_femeninos = ["Mrs", "Miss", "Mlle", "Ms", "Mme", "Lady", "the Countess"]
titulos_masculinos = ["Mr", "Master", "Sir", "Don", "Rev", "Dr", "Major",
                       "Col", "Capt", "Jonkheer"]

def consistencia_titulo_sexo(row):
    if row["titulo"] in titulos_femeninos:
        return row["Sex"] == "female"
    if row["titulo"] in titulos_masculinos:
        return row["Sex"] == "male"
    return True  # título ambiguo (ej: "Dr" podía ser mujer), no se penaliza

df_titanic["titulo_consistente_con_sexo"] = df_titanic.apply(consistencia_titulo_sexo, axis=1)
print("\nRegistros donde el título NO condice con Sex:",
      (~df_titanic["titulo_consistente_con_sexo"]).sum())
````

````python
# ============================================================
# EXPERIMENTACIÓN — IRIS
# ============================================================

# ---------- COMPLETITUD ----------
print(">>> COMPLETITUD")
print(df_iris.isnull().sum())
# Esperable: cero en las 5 columnas.

# ---------- UNICIDAD ----------
print("\n>>> UNICIDAD (H1)")
df_iris["fila_duplicada"] = df_iris.duplicated(keep=False)
print("Filas duplicadas exactas:", df_iris.duplicated().sum())
display(df_iris[df_iris["fila_duplicada"]])

# ---------- VALIDEZ ----------
print("\n>>> VALIDEZ")
columnas_medidas = ["sepal_length", "sepal_width", "petal_length", "petal_width"]

# Las medidas deben ser positivas y dentro de un rango biológicamente plausible
df_iris["medidas_positivas"] = (df_iris[columnas_medidas] > 0).all(axis=1)
df_iris["especie_valida"] = df_iris["species"].str.replace("Iris-", "", regex=False) \
    .isin(["setosa", "versicolor", "virginica"])

print("Filas con alguna medida <= 0:", (~df_iris["medidas_positivas"]).sum())
print("Filas con especie fuera de las 3 esperadas:", (~df_iris["especie_valida"]).sum())

# ---------- CONSISTENCIA ----------
print("\n>>> CONSISTENCIA (H2: formato de etiqueta entre fuentes)")
print("Etiquetas únicas en la versión URL:", df_iris["species"].unique())
print("Etiquetas únicas en la versión seaborn:", df_iris_sns["species"].unique())
prefijo_uniforme = df_iris["species"].str.startswith("Iris-").all()
print("¿Todas las etiquetas de la versión URL llevan 'Iris-'?:", prefijo_uniforme)

# Consistencia biológica: el pétalo no debería ser más largo que el sépalo
df_iris["petalo_menor_que_sepalo"] = df_iris["petal_length"] <= df_iris["sepal_length"]
print("\nFilas donde petal_length > sepal_length (inconsistente):",
      (~df_iris["petalo_menor_que_sepalo"]).sum())
````

### ¿Qué se aprendió sobre datos estructurados y no estructurados?

Los tres datasets son formalmente **datos estructurados**: filas y columnas
bien definidas, con un esquema fijo. Sin embargo, el ejercicio mostró que
"estructurado" no es sinónimo de "limpio" ni de "homogéneo":

- Dentro de un dataset estructurado pueden convivir columnas con información
  **no estructurada** disfrazada de texto simple. `Name` en Titanic es el
  ejemplo más claro: a simple vista es solo una cadena de texto, pero contiene
  información codificada (título, apellido, estado civil) que solo se vuelve
  utilizable si se **parsea** con expresiones regulares. Lo mismo pasa con
  `Cabin`, donde la letra inicial (cubierta) y el número van pegados en un
  mismo campo de texto.
- El grado de "limpieza" de un dataset estructurado depende más de **su origen**
  que de su formato. Iris y Ventas están estructurados igual que Titanic, pero
  provienen de fuentes curadas (un experimento de laboratorio controlado, un
  dataset de ejemplo para Kaggle), mientras que Titanic proviene de un registro
  administrativo real de 1912, con toda la irregularidad humana que eso implica.
- Verificar calidad de datos no es solo correr `isnull()` y `duplicated()`. Las
  columnas de texto libre exigen mirar **consistencia semántica** (¿el título
  coincide con el sexo?, ¿la etiqueta de especie es la misma en todas las
  fuentes?), algo que no aparece con un chequeo automático genérico y que
  requiere conocer el dominio de los datos.
- Por último, generar errores a propósito (como se hizo en Ventas) enseña la
  lección inversa: un dataset puede parecer perfecto en `describe()` y aun así
  esconder problemas si el proceso que lo generó (en este caso, un muestreo
  aleatorio) no se documenta. La calidad de un dato no se certifica solo mirando
  el resultado final, sino entendiendo cómo se construyó.

## ¿Por qué NO entra en filas y columnas como los datasets anteriores?

En Titanic, Iris y Ventas, **cada fila es una unidad atómica y comparable**:
un pasajero, una flor, una transacción. Y **cada columna tiene un significado
fijo y un tipo de dato definido** para todas las filas por igual (`Age` siempre
es una edad numérica, `Fare` siempre es un monto). Ese es precisamente el
requisito de un dato estructurado: un esquema común que se repite fila tras fila.

El comentario de arriba rompe ese esquema en varios frentes:

1. **No tiene longitud ni forma fija.** Un pasajero de Titanic siempre ocupa
   una fila con 12 valores. Este comentario podría haber tenido una línea o
   veinte; no hay un número de "campos" predecible.

2. **Mezcla varios datos distintos dentro de una misma oración**, sin
   delimitador que los separe: la categoría del producto, el tiempo de envío
   esperado, el tiempo de envío real, el estado del embalaje, una intención de
   recompra condicional y una calificación numérica. En un dataset estructurado
   cada uno de esos datos sería **su propia columna**; acá están todos
   entrelazados en prosa.

3. **Para extraer un dato hay que interpretar, no solo leer.** El "3.5/5" se
   puede parsear con una expresión regular, pero "tengo sentimientos
   encontrados" o "andaría mejor de lo que pensé" no son datos listos para una
   columna: son **juicios en lenguaje natural** que requieren análisis de
   sentimiento (¿es una opinión positiva, negativa, mixta?) para convertirse en
   algo tabulable, y ese análisis siempre involucra cierto margen de ambigüedad.

4. **Usa recursos que no tienen equivalente en una celda de Excel**: emojis
   (😑), abreviaturas ("jaja"), sarcasmo implícito, y errores de tipeo. Nada de
   eso tiene un tipo de dato pandas que lo represente sin pérdida de información.

5. **No hay garantía de que dos comentarios contengan los mismos "campos".**
   Otro cliente podría escribir solo "excelente, todo perfecto" sin mencionar
   envío, embalaje ni nota. En Ventas, en cambio, **todas** las filas tienen
   valor en `Quantity` y en `Price per Unit`: la completitud del esquema está
   garantizada por diseño, no depende de qué quiso contar cada persona.
