### Preparación de los Datos y Ajuste de Encabezados
Dado que el dataset original no incluye una fila de nombres de columnas, vamos a recargarlo definiendo correctamente los nombres de los atributos indicados arriba.

```python
import pandas as pd

# Lista de nombres de columnas según la descripción del dataset
column_names = [
    'animal_name', 'hair', 'feathers', 'eggs', 'milk', 'airborne',
    'aquatic', 'predator', 'toothed', 'backbone', 'breathes', 'venomous',
    'fins', 'legs', 'tail', 'domestic', 'catsize', 'type'
]

# Volvemos a leer el archivo indicando que no contiene cabecera y asignamos las columnas
df_clean = pd.read_csv('dataset.csv', names=column_names)

# Mostrar las primeras filas ya con los nombres correctos
display(df_clean.head())
```

### Separación de Características y Variable Objetivo
Excluimos la columna `animal_name` por ser una cadena identificativa irrelevante para la predicción del tipo de animal, y definimos `type` como nuestra variable objetivo (y).

```python
# Características (X) y objetivo (y)
X = df_clean.drop(columns=['animal_name', 'type'])
y = df_clean['type']

print("Dimensiones de X:", X.shape)
print("Dimensiones de y:", y.shape)
```

### Búsqueda de Hiperparámetros con Validación Cruzada (Grid Search)
Usaremos `GridSearchCV` para evaluar el rendimiento variando `criterion` ('gini' y 'entropy') y `max_depth` (de 1 a 10).

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import GridSearchCV, StratifiedKFold

# Definir el clasificador base
clf = DecisionTreeClassifier(random_state=42)

# Definir la malla de parámetros a evaluar
param_grid = {
    'criterion': ['gini', 'entropy'],
    'max_depth': [2, 3, 4, 5, 6, 7, 8, 9, 10, None]
}

# Usamos StratifiedKFold para asegurar que cada pliegue mantenga la proporción de clases
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Configurar GridSearchCV
grid_search = GridSearchCV(estimator=clf, param_grid=param_grid, cv=cv, scoring='accuracy', n_jobs=-1)

# Ajustar el modelo
grid_search.fit(X, y)

# Mostrar mejores parámetros y resultado de precisión (accuracy)
print(f"Mejores parámetros encontrados: {grid_search.best_params_}")
print(f"Mejor precisión promedio de Validación Cruzada: {grid_search.best_score_:.4f}")

# Mostrar el resumen de todos los experimentos en un DataFrame
results_df = pd.DataFrame(grid_search.cv_results_)
print("--- Resultados de las combinaciones evaluadas ---")
display(results_df[['param_criterion', 'param_max_depth', 'mean_test_score', 'std_test_score']].sort_values(by='mean_test_score', ascending=False))
```

### Reporte de métricas
> **Corrección respecto al original:** en la versión original esta celda usaba `y_pred` antes de que existiera (se definía recién en la celda de la matriz de confusión), lo que producía un `NameError`. Ahora las predicciones por validación cruzada se calculan al inicio de la celda.

```python
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import classification_report

# Predicciones por validación cruzada con el mejor árbol (cada fila se predice con un modelo que no la vio)
y_pred = cross_val_predict(grid_search.best_estimator_, X, y, cv=cv)

# Generar el reporte de clasificación detallado usando las predicciones de validación cruzada
report = classification_report(y, y_pred, target_names=[f"Clase {i}" for i in sorted(y.unique())])

print("--- Reporte de Métricas de Clasificación (Validación Cruzada) ---")
print(report)
```

### Visualización del Árbol de Decisión Óptimo
Por último, podemos visualizar gráficamente el árbol resultante con la mejor configuración de parámetros.

```python
import matplotlib.pyplot as plt
from sklearn.tree import plot_tree

# Obtener el mejor estimador entrenado
best_tree = grid_search.best_estimator_

# Configurar y dibujar el árbol
plt.figure(figsize=(20, 10))
plot_tree(
    best_tree,
    feature_names=X.columns,
    class_names=[str(i) for i in sorted(y.unique())],
    filled=True,
    rounded=True,
    fontsize=10
)
plt.title("Árbol de Decisión Óptimo Encontrado por GridSearchCV")
plt.show()
```

### Matriz de Confusión
Utilizamos `cross_val_predict` para obtener las predicciones del mejor modelo en cada pliegue de la validación cruzada y luego generamos la matriz de confusión.

```python
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

# Obtener el mejor estimador entrenado del GridSearchCV
best_tree = grid_search.best_estimator_

# Generar predicciones mediante validación cruzada utilizando el mismo esquema StratifiedKFold (cv)
y_pred = cross_val_predict(best_tree, X, y, cv=cv)

# Calcular la matriz de confusión
cm = confusion_matrix(y, y_pred)

# Graficar la matriz de confusión de forma visual y legible
plt.figure(figsize=(10, 8))
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=sorted(y.unique()))
disp.plot(cmap='Blues', ax=plt.gca())
plt.title("Matriz de Confusión mediante Validación Cruzada")
plt.grid(False)  # Desactivar la cuadrícula sobre el gráfico para mejor visibilidad
plt.show()
```

## 1 - Credit Score
Dataset de clasificación de **puntaje crediticio** (`parisrohan/credit-score-classification`, el mismo de la consigna). Cada fila es un mes de un cliente; la variable objetivo `Credit_Score` tiene tres clases: **Poor**, **Standard** y **Good**.

```python
import kagglehub                                            # librería de descarga desde Kaggle

# Descarga la última versión del dataset y guarda la carpeta local en `path`
path = kagglehub.dataset_download("parisrohan/credit-score-classification")

print("Path to dataset files:", path)                       # muestra la ruta de descarga
print(os.listdir(path))                                     # lista los archivos (train.csv y test.csv)
```

### 1.1 Carga y exploración
Se usa `train.csv` porque es el único que trae la variable objetivo.

```python
raw_credit = pd.read_csv(os.path.join(path, 'train.csv'))   # lee el CSV de entrenamiento (original intocable)
df_credit = raw_credit.copy()                               # copia de trabajo: el original no se modifica

print('Forma (filas, columnas):', df_credit.shape)          # cantidad de filas y columnas
display(df_credit.head())                                   # primeras 5 filas
df_credit.info()                                            # tipos de dato y nulos

print('\nDistribución del objetivo:')                       # título
print(df_credit['Credit_Score'].value_counts(normalize=True).round(3))  # proporción de cada clase
print('\nClientes distintos:', df_credit['Customer_ID'].nunique())      # cada cliente aparece varios meses
```

### 1.2 Limpieza de datos
Este dataset viene "sucio" a propósito. Problemas típicos y cómo se tratan:

| Problema | Ejemplo | Tratamiento |
|---|---|---|
| Números guardados como texto con `_` | `'23_'`, `'19114.12_'` | quitar `_` y convertir con `to_numeric` |
| Antigüedad crediticia como texto | `'22 Years and 1 Months'` | convertir a **meses** (numérico) |
| Marcadores de dato inválido | `'_______'`, `'_'`, `'!@9#%8'` | reemplazar por `NaN` |
| Valores imposibles | edad negativa o > 100, préstamos negativos | pasar a `NaN` |
| Columnas identificadoras o de texto libre | `ID`, `Name`, `SSN`, `Type_of_Loan` | eliminar de las predictoras |

Los `NaN` resultantes se imputan **dentro del Pipeline** (solo con datos de entrenamiento de cada pliegue), así la validación cruzada no se contamina.

```python
# --- 1) Columnas numéricas guardadas como texto con guiones bajos ---
cols_texto_num = ['Age', 'Annual_Income', 'Num_of_Loan', 'Num_of_Delayed_Payment', 'Changed_Credit_Limit',
                  'Outstanding_Debt', 'Amount_invested_monthly', 'Monthly_Balance']   # columnas a convertir
for c in cols_texto_num:                                    # recorre cada columna
    limpio = df_credit[c].astype(str).str.replace('_', '', regex=False).str.strip()   # quita '_' y espacios
    df_credit[c] = pd.to_numeric(limpio, errors='coerce')   # convierte a número (lo inválido pasa a NaN)

# --- 2) Antigüedad crediticia: '22 Years and 1 Months' -> total de meses ---
anios = df_credit['Credit_History_Age'].astype(str).str.extract(r'(\d+)\s*Years')[0].astype(float)   # años
meses = df_credit['Credit_History_Age'].astype(str).str.extract(r'(\d+)\s*Months')[0].astype(float)  # meses
df_credit['Credit_History_Months'] = anios * 12 + meses     # total en meses

# --- 3) Marcadores de dato inválido en categóricas -> NaN ---
df_credit['Occupation'] = df_credit['Occupation'].replace('_______', np.nan)          # ocupación inválida
df_credit['Credit_Mix'] = df_credit['Credit_Mix'].replace('_', np.nan)                # mezcla de crédito inválida
df_credit['Payment_Behaviour'] = df_credit['Payment_Behaviour'].replace('!@9#%8', np.nan)  # comportamiento inválido

# --- 4) Valores imposibles según reglas de negocio -> NaN ---
df_credit.loc[(df_credit['Age'] < 14) | (df_credit['Age'] > 100), 'Age'] = np.nan     # edad fuera de rango
df_credit.loc[df_credit['Num_of_Loan'] < 0, 'Num_of_Loan'] = np.nan                   # préstamos negativos
df_credit.loc[df_credit['Num_of_Delayed_Payment'] < 0, 'Num_of_Delayed_Payment'] = np.nan  # atrasos negativos
df_credit.loc[df_credit['Monthly_Balance'].abs() > 1e6, 'Monthly_Balance'] = np.nan   # saldo con valor absurdo

# --- 5) Verificación ---
print('Nulos por columna (solo las que tienen):')
nulos = df_credit.isnull().sum()                            # cuenta nulos por columna
print(nulos[nulos > 0])                                     # muestra solo las que tienen
display(df_credit.describe().T)                             # descriptivos numéricos tras la limpieza
```

### 1.3 Separación de características (X) y objetivo (y)
Se excluyen las columnas identificadoras (`ID`, `Customer_ID`, `Name`, `SSN`), el mes, el texto libre `Type_of_Loan` y la columna original `Credit_History_Age` (ya convertida).

**Detalle importante:** cada cliente aparece ~8 veces (una por mes) con valores muy parecidos. Si el mismo cliente cae en el set de entrenamiento y en el de prueba, la validación se vuelve optimista. Por eso se usa `StratifiedGroupKFold` agrupando por `Customer_ID`: todos los meses de un cliente quedan juntos en el mismo pliegue.

```python
cols_excluir = ['ID', 'Customer_ID', 'Name', 'SSN', 'Month', 'Type_of_Loan',
                'Credit_History_Age', 'Credit_Score']       # columnas que NO son predictoras

X_cs = df_credit.drop(columns=cols_excluir)                 # características
y_cs = df_credit['Credit_Score']                            # variable objetivo (Poor / Standard / Good)
grupos_cs = df_credit['Customer_ID']                        # cliente de cada fila (para no mezclar clientes entre pliegues)

num_cols_cs = X_cs.select_dtypes(include='number').columns.tolist()   # columnas numéricas
cat_cols_cs = X_cs.select_dtypes(exclude='number').columns.tolist()   # columnas categóricas

print('X:', X_cs.shape, '| y:', y_cs.shape)                 # dimensiones
print('Numéricas:', len(num_cols_cs), '| Categóricas:', cat_cols_cs)  # resumen de tipos
```

### 1.4 Árbol de decisión con validación cruzada (GridSearchCV)
- **Pipeline:** imputación de nulos (mediana en numéricas, moda en categóricas) + OneHot en categóricas + árbol. Los árboles no necesitan escalado.
- **Grilla:** `criterion` (gini / entropy), `max_depth` y `min_samples_leaf`. Limitar la profundidad evita el sobreajuste.
- **Validación cruzada:** 5 pliegues, con grupos por cliente y estratificación por clase.
- **Métrica de selección:** `f1_macro` (trata igual a las tres clases).

```python
# Preprocesamiento: cada tipo de columna recibe su tratamiento
preprocesador_cs = ColumnTransformer([
    ('num', SimpleImputer(strategy='median'), num_cols_cs),                 # numéricas: nulos -> mediana
    ('cat', Pipeline([
        ('imp', SimpleImputer(strategy='most_frequent')),                   # categóricas: nulos -> moda
        ('ohe', OneHotEncoder(handle_unknown='ignore', sparse_output=False))  # a columnas 0/1
    ]), cat_cols_cs)
])

# Pipeline completo: preprocesar y luego clasificar
pipe_cs = Pipeline([
    ('prep', preprocesador_cs),                             # paso 1: preprocesamiento
    ('tree', DecisionTreeClassifier(random_state=42))       # paso 2: árbol de decisión
])

# Malla de hiperparámetros (el prefijo 'tree__' indica que pertenecen al paso 'tree')
param_grid_cs = {
    'tree__criterion': ['gini', 'entropy'],                 # medida de impureza
    'tree__max_depth': [3, 5, 7, 10, 15, None],             # profundidad máxima
    'tree__min_samples_leaf': [1, 20]                       # mínimo de muestras por hoja
}

# Esquema de validación cruzada: 5 pliegues, estratificado y agrupado por cliente
cv_cs = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=42)

# Búsqueda en grilla
grid_cs = GridSearchCV(pipe_cs, param_grid_cs, cv=cv_cs, scoring='f1_macro', n_jobs=-1)
grid_cs.fit(X_cs, y_cs, groups=grupos_cs)                   # ajusta (groups indica el cliente de cada fila)

print('Mejores parámetros:', grid_cs.best_params_)          # combinación ganadora
print(f'Mejor F1-macro (validación cruzada): {grid_cs.best_score_:.4f}')   # puntaje promedio

res_cs = pd.DataFrame(grid_cs.cv_results_)                  # tabla con todas las combinaciones
display(res_cs[['param_tree__criterion', 'param_tree__max_depth', 'param_tree__min_samples_leaf',
                'mean_test_score', 'std_test_score']].sort_values('mean_test_score', ascending=False).head(10))
```

### 1.5 Métricas de la validación cruzada
Se evalúa el mejor árbol en cada pliegue y se informa el **promedio ± desvío** de varias métricas. `macro` promedia las clases por igual; `weighted` pondera por cantidad de casos.

```python
mejor_cs = grid_cs.best_estimator_                          # mejor pipeline encontrado

# Métricas a calcular en cada pliegue
metricas = {'accuracy': 'accuracy', 'precision_macro': 'precision_macro', 'recall_macro': 'recall_macro',
            'f1_macro': 'f1_macro', 'f1_weighted': 'f1_weighted'}

# Validación cruzada con múltiples métricas (usa los mismos pliegues agrupados por cliente)
cv_res_cs = cross_validate(mejor_cs, X_cs, y_cs, cv=cv_cs, groups=grupos_cs, scoring=metricas, n_jobs=-1)

print('--- Métricas por validación cruzada (promedio ± desvío) ---')
for nombre in metricas:                                     # recorre cada métrica
    valores = cv_res_cs['test_' + nombre]                   # valor en cada pliegue
    print(f'{nombre:16s}: {valores.mean():.4f} ± {valores.std():.4f}')

# Predicciones fuera de muestra para el reporte por clase y la matriz de confusión
y_pred_cs = cross_val_predict(mejor_cs, X_cs, y_cs, cv=cv_cs, groups=grupos_cs, n_jobs=-1)

print('\n--- Reporte de clasificación por clase ---')
print(classification_report(y_cs, y_pred_cs, digits=3))     # precision, recall y f1 de cada clase

# Guarda el resumen para la comparación final
resumen_resultados['Credit Score'] = {
    'mejores_parametros': grid_cs.best_params_,
    'accuracy': cv_res_cs['test_accuracy'].mean(),
    'f1_macro': cv_res_cs['test_f1_macro'].mean()}
```

### 1.6 Matriz de confusión

```python
orden_cs = ['Poor', 'Standard', 'Good']                     # orden lógico de las clases

fig, ejes = plt.subplots(1, 2, figsize=(14, 5))             # dos paneles: cantidades y proporciones

# Panel 1: matriz con cantidades absolutas
ConfusionMatrixDisplay.from_predictions(y_cs, y_pred_cs, labels=orden_cs, cmap='Blues',
                                        ax=ejes[0], colorbar=False)
ejes[0].set_title('Matriz de confusión (cantidades)')       # título
ejes[0].grid(False)                                         # sin cuadrícula

# Panel 2: matriz normalizada por fila (cada fila suma 1 = recall de la clase)
ConfusionMatrixDisplay.from_predictions(y_cs, y_pred_cs, labels=orden_cs, cmap='Blues',
                                        normalize='true', values_format='.2f', ax=ejes[1], colorbar=False)
ejes[1].set_title('Matriz de confusión (proporción por clase real)')
ejes[1].grid(False)

plt.tight_layout()                                          # evita superposición
plt.show()                                                  # muestra el gráfico
```

### 1.7 Árbol e importancia de variables
Se dibujan solo los primeros 3 niveles para que sea legible (el árbol completo puede tener cientos de nodos).

```python
arbol_cs = mejor_cs.named_steps['tree']                     # el árbol dentro del pipeline
nombres_cs = mejor_cs.named_steps['prep'].get_feature_names_out()   # nombres de las columnas tras el preprocesamiento

plt.figure(figsize=(22, 9))                                 # tamaño grande para leer los nodos
plot_tree(arbol_cs, feature_names=nombres_cs, class_names=[str(c) for c in arbol_cs.classes_],
          filled=True, rounded=True, max_depth=3, fontsize=9)   # dibuja hasta profundidad 3
plt.title('Árbol de decisión - Credit Score (primeros 3 niveles)')
plt.show()

# Importancia de variables: cuánto aporta cada una a reducir la impureza
imp_cs = pd.Series(arbol_cs.feature_importances_, index=nombres_cs).sort_values(ascending=False).head(15)
plt.figure(figsize=(8, 5))                                  # gráfico de barras
sns.barplot(x=imp_cs.values, y=imp_cs.index, color='steelblue')
plt.title('Top 15 variables más importantes - Credit Score')
plt.xlabel('Importancia')
plt.tight_layout()
plt.show()
```

## 2 - Stellar Classification (SDSS17)
Dataset del *Sloan Digital Sky Survey* (`fedesoriano/stellar-classification-dataset-sdss17`): 100.000 observaciones de objetos astronómicos. La variable objetivo `class` tiene tres clases: **GALAXY**, **STAR** y **QSO** (cuásar).

Variables principales: `alpha`/`delta` (coordenadas), `u, g, r, i, z` (brillo en cinco filtros fotométricos) y `redshift` (corrimiento al rojo).

```python
import kagglehub                                            # librería de descarga desde Kaggle

# Descarga el dataset completo y guarda la ruta local
path_st = kagglehub.dataset_download("fedesoriano/stellar-classification-dataset-sdss17")
print("Ruta del dataset:", path_st)

# Busca los archivos .csv dentro de la carpeta descargada
archivos_st = os.listdir(path_st)                           # todos los archivos
csv_st = [f for f in archivos_st if f.endswith('.csv')]     # solo los CSV
print("Archivos encontrados:", archivos_st)
```

### 2.1 Carga y exploración

```python
raw_stellar = pd.read_csv(os.path.join(path_st, csv_st[0]))   # lee el primer CSV encontrado (original intocable)
df_stellar = raw_stellar.copy()                               # copia de trabajo

print('Forma (filas, columnas):', df_stellar.shape)         # dimensiones
display(df_stellar.head())                                  # primeras filas
df_stellar.info()                                           # tipos y nulos
display(df_stellar.describe().T)                            # descriptivos (ojo con los mínimos de u, g, z)

print('\nDistribución de la variable objetivo:')
print(df_stellar['class'].value_counts(normalize=True).round(3))   # proporción de cada clase
print('Filas duplicadas:', df_stellar.duplicated().sum())   # unicidad
```

### 2.2 Limpieza de datos
- **Valores centinela:** en este dataset hay un registro con `-9999` en `u`, `g` y `z` (valor que indica "medición fallida"). Se elimina esa fila.
- **Identificadores:** `obj_ID`, `run_ID`, `rerun_ID`, `cam_col`, `field_ID`, `spec_obj_ID`, `plate`, `MJD` y `fiber_ID` identifican la observación o el instrumento, no propiedades del objeto; se excluyen para que el árbol no aprenda "marcas" de la toma de datos.

```python
# Detecta valores centinela (-9999) en los filtros fotométricos
centinela = (df_stellar[['u', 'g', 'z']] <= -9000).any(axis=1)   # True si alguno vale -9999
print('Filas con valor centinela:', centinela.sum())              # cuántas hay

df_stellar = df_stellar[~centinela].reset_index(drop=True)        # las elimina y reinicia el índice

cols_id = ['obj_ID', 'run_ID', 'rerun_ID', 'cam_col', 'field_ID',
           'spec_obj_ID', 'plate', 'MJD', 'fiber_ID']             # columnas identificadoras
X_st = df_stellar.drop(columns=cols_id + ['class'])               # características físicas
y_st = df_stellar['class']                                        # objetivo: GALAXY / STAR / QSO

print('Predictoras:', list(X_st.columns))                         # columnas que usará el árbol
print('X:', X_st.shape, '| y:', y_st.shape)                       # dimensiones
print('Nulos en X:', int(X_st.isnull().sum().sum()))              # debe ser 0
```

### 2.3 Árbol de decisión con validación cruzada (GridSearchCV)
Todas las variables son numéricas y no hay nulos, por lo que no hace falta preprocesamiento. Se usa `StratifiedKFold` de 5 pliegues porque las clases están desbalanceadas (~59 % GALAXY, ~22 % STAR, ~19 % QSO) y la métrica de selección es `f1_macro`.

```python
# Esquema de validación cruzada: 5 pliegues manteniendo la proporción de clases
cv_st = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# Malla de hiperparámetros
param_grid_st = {
    'criterion': ['gini', 'entropy'],                       # medida de impureza
    'max_depth': [3, 5, 7, 10, 15, None],                   # profundidad máxima
    'min_samples_leaf': [1, 20]                             # mínimo de muestras por hoja
}

# Búsqueda en grilla con el árbol base
grid_st = GridSearchCV(DecisionTreeClassifier(random_state=42), param_grid_st,
                       cv=cv_st, scoring='f1_macro', n_jobs=-1)
grid_st.fit(X_st, y_st)                                     # ajusta todas las combinaciones

print('Mejores parámetros:', grid_st.best_params_)          # combinación ganadora
print(f'Mejor F1-macro (validación cruzada): {grid_st.best_score_:.4f}')

res_st = pd.DataFrame(grid_st.cv_results_)                  # tabla con todas las combinaciones
display(res_st[['param_criterion', 'param_max_depth', 'param_min_samples_leaf',
                'mean_test_score', 'std_test_score']].sort_values('mean_test_score', ascending=False).head(10))
```

### 2.4 Métricas de la validación cruzada

```python
mejor_st = grid_st.best_estimator_                          # mejor árbol encontrado

# Validación cruzada con múltiples métricas (reutiliza las de la sección 1)
cv_res_st = cross_validate(mejor_st, X_st, y_st, cv=cv_st, scoring=metricas, n_jobs=-1)

print('--- Métricas por validación cruzada (promedio ± desvío) ---')
for nombre in metricas:                                     # recorre cada métrica
    valores = cv_res_st['test_' + nombre]                   # valor en cada pliegue
    print(f'{nombre:16s}: {valores.mean():.4f} ± {valores.std():.4f}')

# Predicciones fuera de muestra para el reporte por clase y la matriz de confusión
y_pred_st = cross_val_predict(mejor_st, X_st, y_st, cv=cv_st, n_jobs=-1)

print('\n--- Reporte de clasificación por clase ---')
print(classification_report(y_st, y_pred_st, digits=3))     # precision, recall y f1 de cada clase

# Guarda el resumen para la comparación final
resumen_resultados['Stellar'] = {
    'mejores_parametros': grid_st.best_params_,
    'accuracy': cv_res_st['test_accuracy'].mean(),
    'f1_macro': cv_res_st['test_f1_macro'].mean()}
```

### 2.5 Matriz de confusión
