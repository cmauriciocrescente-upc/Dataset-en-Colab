##2-Inventario de Archivos:

```python
import os
from pathlib import Path
from collections import Counter

RUTA = Path('.')   # cambiá si usaste Drive, ej: Path('/content/drive/MyDrive/tu_carpeta')

archivos = [p for p in RUTA.rglob('*') if p.is_file() and '.git' not in p.parts]
print(f"Total de archivos: {len(archivos)}\n")

ext = Counter(p.suffix.lower() or '(sin extensión)' for p in archivos)
for e, n in ext.most_common():
    print(f"{e:20s} {n}")

print("\nEstructura de carpetas:")
for p in sorted({p.parent for p in archivos}):
    print(" ", p)
```

##3-Leer cada tipo de archivo:

```python
import pandas as pd
import json

datos = {}      # DataFrames
notebooks = []  # rutas de .ipynb
scripts = {}    # código .py
textos = {}     # README, txt, md

for p in archivos:
    s = p.suffix.lower()
    try:
        if s == '.csv':
            datos[str(p)] = pd.read_csv(p)
        elif s in ('.xlsx', '.xls'):
            datos[str(p)] = pd.read_excel(p)
        elif s == '.json':
            with open(p, encoding='utf-8') as f:
                datos[str(p)] = pd.json_normalize(json.load(f))
        elif s == '.ipynb':
            notebooks.append(p)
        elif s == '.py':
            scripts[str(p)] = p.read_text(encoding='utf-8', errors='ignore')
        elif s in ('.txt', '.md'):
            textos[str(p)] = p.read_text(encoding='utf-8', errors='ignore')
    except Exception as e:
        print(f"⚠️ No se pudo leer {p}: {e}")

print(f"CSV/Excel/JSON cargados: {len(datos)}")
print(f"Notebooks: {len(notebooks)} | Scripts .py: {len(scripts)} | Textos: {len(textos)}")
```

##4-Resumen de cada dataset:

```python
for nombre, df in datos.items():
    print("=" * 70)
    print(nombre)
    print("Filas x columnas:", df.shape)
    print("Nulos totales:", int(df.isna().sum().sum()))
    display(df.head(3))
    display(df.describe(include='all').T.head(10))
```

##5-Extraer el contenido de los notebooks (código y markdown):

```python
for nb_path in notebooks:
    with open(nb_path, encoding='utf-8') as f:
        nb = json.load(f)
    celdas = nb.get('cells', [])
    n_code = sum(c['cell_type'] == 'code' for c in celdas)
    n_md = sum(c['cell_type'] == 'markdown' for c in celdas)
    print(f"📓 {nb_path} → {n_code} celdas de código, {n_md} de markdown")

    # Para ver el título/primeras líneas de markdown:
    for c in celdas:
        if c['cell_type'] == 'markdown':
            print("   ", ''.join(c['source'])[:100].replace('\n', ' '))
            break
```

##6-Ver librerías que usan los proyectos:

```python
import re
librerias = Counter()
patron = re.compile(r'^\s*(?:import|from)\s+([a-zA-Z0-9_]+)', re.M)

for codigo in scripts.values():
    librerias.update(patron.findall(codigo))

for nb_path in notebooks:
    with open(nb_path, encoding='utf-8') as f:
        nb = json.load(f)
    for c in nb.get('cells', []):
        if c['cell_type'] == 'code':
            librerias.update(patron.findall(''.join(c['source'])))

print(librerias.most_common(20))
```
